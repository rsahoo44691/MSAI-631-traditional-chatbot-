```python
import nltk
from nltk.chat.util import Chat, reflections
import re
import random

nltk.download('punkt', quiet=True)
print("Setup complete. Reflections available:", len(reflections), "pronoun swaps")
```

```python
BOT_NAME = "HelpBot"
BOT_VERSION = "1.0"

CAPABILITIES = [
      "Greet you (try: hi, hello)",
      "Tell you what I can do (try: help, what can you do)",
      "Tell you my name (try: who are you, what is your name)",
      "Tell the current time (try: what time is it)",
      "Tell a joke (try: tell me a joke)",
      "Do basic small talk (try: how are you)",
      "Say goodbye (try: bye, quit, exit)",
  ]
JOKES = [
      "Why did the developer go broke? Because he used up all his cache.",
      "Why do Python programmers wear glasses? Because they can't C.",
      "I told my computer I needed a break — it said 'no problem, I'll go to sleep.'",
  ]

def list_capabilities():
      return "Here's what I can do:\n- " + "\n- ".join(CAPABILITIES)

def current_time():
      from datetime import datetime
      return "The current time is " + datetime.now().strftime("%I:%M %p")
```

```python
pairs = [
      [r"(hi|hello|hey|hola)\b.*",
       ["Hello!", "Hi there!", "Hey, nice to meet you!"]],

      [r".*(help|what can you do|capabilities|commands).*",
       [list_capabilities()]],

      [r".*(your name|who are you).*",
       [f"I'm {BOT_NAME}, version {BOT_VERSION}. A traditional rule-based chatbot."]],

      [r".*(time).*",
       [current_time()]],

      [r".*(joke|funny|make me laugh).*",
       [random.choice(JOKES) for _ in range(3)]],

      [r".*how are you.*",
       ["I'm running on rules and loving it!", "All systems nominal — how about you?"]],

      [r".*(thank you|thanks).*",
       ["You're welcome!", "Happy to help."]],

      [r"(bye|quit|exit|goodbye).*",
       ["Goodbye!", "See you later!"]],
  ]
chatbot = Chat(pairs, reflections)
```

```python
FALLBACK = (
    "I didn't understand that. Try asking for 'help' to see what I can do, "
    "or rephrase your question."
)

def get_response(user_input: str) -> str:

    if not user_input or not user_input.strip():
        return "It looks like you didn't type anything. Try saying 'hello' or 'help'."

    cleaned = user_input.strip().lower()
    response = chatbot.respond(cleaned)

    if response is None:
        # Hook for next week: call Azure Cognitive Services here for unknown input.
        return FALLBACK
    return response
```

```python
def run_chat():
    print(f"{BOT_NAME} v{BOT_VERSION}: Hi! Type 'help' to see what I can do, or 'bye' to quit.\n")
    while True:
        user = input("You: ")
        reply = get_response(user)
        print(f"{BOT_NAME}: {reply}\n")
        if re.match(r"(bye|quit|exit|goodbye)", user.strip().lower()):
            break

run_chat()
```
